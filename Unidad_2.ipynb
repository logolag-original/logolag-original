# Convexity and quadratic functions

En matemáticas, un conjunto $S$ en un espacio vectorial real se llama **convexo** si, para cada par de puntos $x$ e $y$ en $S$, el segmento de línea que los une también está contenido completamente en $S$. En otras palabras, si $x, y \in S$, entonces para cualquier $\alpha \in [0, 1]$, el punto $(1 - \alpha)x + \alpha y$ también está en $S$.

Ejemplos comunes de conjuntos convexos en un plano incluyen círculos, elipses, cuadrados, rectángulos y triángulos.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Polygon, Circle

plt.figure(figsize=(10, 5))

# --- Conjunto Convexo 1: Un círculo ---
ax1 = plt.subplot(1, 2, 1)
circle = Circle((0.5, 0.5), 0.4, color='blue', alpha=0.5, label='Círculo (Convexo)')
ax1.add_patch(circle)
ax1.set_xlim(0, 1)
ax1.set_ylim(0, 1)
ax1.set_aspect('equal', adjustable='box')
ax1.set_title('Ejemplo 1: Círculo (Convexo)')
ax1.grid(True)

# --- Conjunto Convexo 2: Un Polígono (Triángulo) ---
ax2 = plt.subplot(1, 2, 2)
points = np.array([[0.2, 0.2], [0.8, 0.3], [0.5, 0.8]])
triangle = Polygon(points, closed=True, color='green', alpha=0.5, label='Triángulo (Convexo)')
ax2.add_patch(triangle)
ax2.set_xlim(0, 1)
ax2.set_ylim(0, 1)
ax2.set_aspect('equal', adjustable='box')
ax2.set_title('Ejemplo 2: Triángulo (Convexo)')
ax2.grid(True)

plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 5))

# --- Conjunto Convexo 3: Un Rectángulo ---
ax3 = plt.subplot(1, 2, 1)
rectangle_coords = np.array([[0.1, 0.1], [0.9, 0.1], [0.9, 0.7], [0.1, 0.7]])
rectangle = Polygon(rectangle_coords, closed=True, color='red', alpha=0.5, label='Rectángulo (Convexo)')
ax3.add_patch(rectangle)
ax3.set_xlim(0, 1)
ax3.set_ylim(0, 1)
ax3.set_aspect('equal', adjustable='box')
ax3.set_title('Ejemplo 3: Rectángulo (Convexo)')
ax3.grid(True)

# --- Conjunto No Convexo (ejemplo visual) ---
ax4 = plt.subplot(1, 2, 2)
non_convex_points = np.array([[0.1, 0.1], [0.9, 0.1], [0.5, 0.5], [0.9, 0.9], [0.1, 0.9]])
non_convex = Polygon(non_convex_points, closed=True, color='purple', alpha=0.5, label='Forma de Estrella (No Convexo)')
ax4.add_patch(non_convex)

# Dibujar un segmento de línea que sale del conjunto para ilustrar la no convexidad
ax4.plot([0.15, 0.85], [0.5, 0.5], 'k--', linewidth=2, label='Segmento fuera del conjunto')
ax4.plot([0.15, 0.85], [0.5, 0.5], 'ko', markersize=8)

ax4.set_xlim(0, 1)
ax4.set_ylim(0, 1)
ax4.set_aspect('equal', adjustable='box')
ax4.set_title('Ejemplo 4: Forma No Convexa')
ax4.grid(True)
ax4.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

## Composición de funciones

En matemáticas, la **composición de funciones** es la aplicación de una función al resultado de otra función. Si tenemos dos funciones, `f(x)` y `g(x)`, la composición `f(g(x))` (leída como 'f de g de x') significa que primero aplicamos la función `g` a `x`, y luego aplicamos la función `f` al resultado de `g(x)`.

En Python, podemos lograr esto simplemente anidando las llamadas a las funciones.

example 1

```python
import sympy
import numpy as np # Keep numpy for lambdify backend
import matplotlib.pyplot as plt # Import matplotlib for plotting

from sympy import symbols, Function, log, sqrt # Import sympy's log and sqrt

# 1. Definir el símbolo para la variable independiente
x_sym = symbols('x')


# 2. Definir las funciones simbólicamente
f_sym = -1 * log(x_sym)  # Usar sympy.log

g_sym = log(x_sym) * sqrt(x_sym)  # Usar sympy.log y sympy.sqrt

print(f"Función f(x) = -ln(x): {f_sym}")
print(f"Función g(x) = ln(x) * sqrt(x): {g_sym}")

# 3. Realizar la composición de funciones simbólicamente

# Composición 1: f(g(x)) - primero g(x), luego f sobre el resultado
# Esto significa sustituir 'x' en f_sym por g_sym
composition_fg_sym = f_sym.subs(x_sym, g_sym)
print(f"\nComposición f(g(x)): {composition_fg_sym}")

# Composición 2: g(f(x)) - primero f(x), luego g sobre el resultado
# Esto significa sustituir 'x' en g_sym por f_sym
composition_gf_sym = g_sym.subs(x_sym, f_sym)
print(f"Composición g(f(x)): {composition_gf_sym}")

# --- Opcional: convertir a funciones Python callable para evaluación numérica ---
from sympy.utilities.lambdify import lambdify

# Convertir las composiciones simbólicas a funciones Python
composition_fg_python = lambdify(x_sym, composition_fg_sym, 'numpy')
composition_gf_python = lambdify(x_sym, composition_gf_sym, 'numpy')

# Ejemplo de uso con valores numéricos adecuados para cada composición
input_fg_value = 2.0  # For f(g(x)), domain requires x > 1
input_gf_value = 0.5  # For g(f(x)), domain requires 0 < x < 1

result_fg = composition_fg_python(input_fg_value)
print(f"\nEvaluando f(g({input_fg_value})): {result_fg}")

result_gf = composition_gf_python(input_gf_value)
print(f"Evaluando g(f({input_gf_value})): {result_gf}")



# Plotting the function g(f(x))
fig = plt.figure(figsize=(8, 6))
ax = plt.axes()
# Define x range for plotting g(f(x)) within its valid domain (0 < x < 1)
x_plot_gf = np.linspace(0.1, 0.99, 100)
ax.plot(x_plot_gf, composition_gf_python(x_plot_gf), label='g(f(x))')
ax.legend()
plt.title('Plot of g o f') # Adding a title for clarity

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('g(f(x))')
plt.grid(True)
plt.show()
```

### Composición y Gráfico de Nuevas Funciones Multivariables

Ahora, vamos a generar la composición `g(f(x,y))` y graficarla en 3D para las siguientes funciones:

*   `f(x,y) = e^(x + y)`
*   `g(x,y) = y^3 * x^2 + y^2`

```python
import sympy
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Definir los símbolos para las variables independientes
x_sym, y_sym = sympy.symbols('x y')

# 2. Definir las nuevas funciones simbólicamente
f_new_sym = sympy.exp(x_sym + y_sym)
# Asumiendo 'y3' era un typo y se refiere a y^3
g_new_sym = y_sym**3 * x_sym**2 + y_sym**2

print(f"Nueva Función f(x,y) = e^(x+y): {f_new_sym}")
print(f"Nueva Función g(x,y) = y^3 * x^2 + y^2: {g_new_sym}")

# 3. Realizar la composición g(f(x,y))
# Esto significa sustituir x_sym por f_new_sym y y_sym por f_new_sym en g_new_sym
composition_gf_new_sym = g_new_sym.subs({x_sym: f_new_sym, y_sym: f_new_sym})
print(f"\nComposición g(f(x,y)): {composition_gf_new_sym}")

# --- Opcional: convertir a una función Python callable para evaluación numérica ---
from sympy.utilities.lambdify import lambdify

# Convertir la composición simbólica a una función Python que acepte arrays de numpy
composition_gf_new_python = lambdify([x_sym, y_sym], composition_gf_new_sym, 'numpy')

# 4. Preparar datos para la gráfica 3D
x_vals = np.linspace(-1, 1, 50)  # Rango para x
y_vals = np.linspace(-1, 1, 50)  # Rango para y
X_grid, Y_grid = np.meshgrid(x_vals, y_vals)

# Calcular los valores de Z para la composición
Z_composed = composition_gf_new_python(X_grid, Y_grid)

# 5. Crear la gráfica 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Trazar la superficie
surface = ax.plot_surface(X_grid, Y_grid, Z_composed, cmap='viridis', alpha=0.8)

# Añadir etiquetas y título
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('g(f(x,y))')
ax.set_title('Gráfica de la Composición g(f(x,y))')

# Añadir barra de color
fig.colorbar(surface, shrink=0.5, aspect=10)

plt.show()
```

### Verificación de la Desigualdad de Jensen para una Variable Aleatoria de Dos Estados

Aquí implementamos una función para verificar la desigualdad de Jensen (`f(E[X]) ≤ E[f(X)]`) para una variable aleatoria que toma dos valores, `x1` y `x2`, con probabilidades `p` y `1-p` respectivamente.

```python
import numpy as np

def verify_jensen_inequality(x1, x2, p, f_convex):
    """
    Verifica la desigualdad de Jensen para una variable aleatoria de dos estados.

    Args:
        x1 (float): El primer valor que puede tomar la variable aleatoria.
        x2 (float): El segundo valor que puede tomar la variable aleatoria.
        p (float): La probabilidad de que la variable aleatoria tome el valor x1 (0 <= p <= 1).
        f_convex (callable): Una función convexa f(x).

    Returns:
        bool: True si f(E[X]) <= E[f(X]), False en caso contrario.
        tuple: (f_ex, ef_x), los valores de f(E[X]) y E[f(X)] respectivamente.
    """
    if not (0 <= p <= 1):
        raise ValueError("La probabilidad 'p' debe estar entre 0 y 1.")

    # Calcular la esperanza (media) de X: E[X]
    expected_value_x = p * x1 + (1 - p) * x2

    # Calcular f(E[X])
    f_of_expected_value_x = f_convex(expected_value_x)

    # Calcular E[f(X)]
    expected_value_f_of_x = p * f_convex(x1) + (1 - p) * f_convex(x2)

    print(f"\nValores de entrada: x1={x1}, x2={x2}, p={p}")
    print(f"E[X] = {expected_value_x:.4f}")
    print(f"f(E[X]) = {f_of_expected_value_x:.4f}")
    print(f"E[f(X)] = {expected_value_f_of_x:.4f}")

    # Verificar la desigualdad
    is_jensen_true = f_of_expected_value_x <= expected_value_f_of_x
    print(f"¿Se cumple f(E[X]) <= E[f(X])? {is_jensen_true}")

    return is_jensen_true, (f_of_expected_value_x, expected_value_f_of_x)

# --- Ejemplo de uso ---
# Definir una función convexa (ejemplo: f(x) = x^2)
def convex_function_square(x):
    return x**2

# Definir una función convexa (ejemplo: f(x) = exp(x))
def convex_function_exp(x):
    return np.exp(x)

# Caso 1: f(x) = x^2, x1=1, x2=5, p=0.5
print("\n--- Caso 1: Función f(x) = x^2 ---")
_ , (f_ex1, ef_x1) = verify_jensen_inequality(x1=1, x2=5, p=0.5, f_convex=convex_function_square)

# Caso 2: f(x) = x^2, x1=0, x2=10, p=0.2
print("\n--- Caso 2: Función f(x) = x^2 ---")
_ , (f_ex2, ef_x2) = verify_jensen_inequality(x1=0, x2=10, p=0.2, f_convex=convex_function_square)

# Caso 3: f(x) = exp(x), x1=-1, x2=1, p=0.7
print("\n--- Caso 3: Función f(x) = exp(x) ---")
_ , (f_ex3, ef_x3) = verify_jensen_inequality(x1=-1, x2=1, p=0.7, f_convex=convex_function_exp)
```

**Quadratic systems (forms)**


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
def compute_quadratic(P, q, r, x):
    """
    Evaluates a quadratic function f(x) = x^T P x + q^T x + r.
    Inputs:
    - P: A 2D numpy array (matrix) for the quadratic term.
    - q: A 1D numpy array (vector) for the linear term.
    - r: A scalar for the constant term.
    - x: A 1D numpy array (vector) at which the quadratic function is evaluated.

    Returns:
    - f(x): The scalar value of the quadratic function.
    """
    # Calculate the quadratic term: x^T * P * x
    quadratic_term = np.dot(x.T, np.dot(P, x))
    # Calculate the linear term: q^T * x
    linear_term = np.dot(q.T, x)
    # Combine all terms
    return quadratic_term + linear_term + r
```

```python
import sympy
import numpy as np

def get_quadratic_form_coeffs(sym_func, sym_vars):
    """
    Extrae la matriz P, el vector q y el escalar r de una función cuadrática simbólica.
    Asume que la función es de la forma f(x) = x^T P x + q^T x + r.

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función cuadrática.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        tuple: (P, q, r) como numpy arrays o escalar.
    """
    num_vars = len(sym_vars)

    # 1. Calcular r (término constante)
    # Evaluar la función en el origen (todas las variables = 0)
    subs_dict = {var: 0 for var in sym_vars}
    r = float(sym_func.subs(subs_dict))

    # 2. Calcular q (vector lineal)
    # El gradiente de f(x) en x=0 nos da q
    gradient = [sympy.diff(sym_func, var) for var in sym_vars]
    q_list = [float(expr.subs(subs_dict)) for expr in gradient]
    q = np.array(q_list)

    # 3. Calcular P (matriz cuadrática)
    # La matriz P es 0.5 * la Matriz Hessiana de f(x)
    hessian = sympy.hessian(sym_func, sym_vars)
    P_sympy = 0.5 * hessian
    P = np.array(P_sympy.tolist(), dtype=float) # Convertir la matriz de sympy a numpy

    return P, q, r

# --- Ejemplo de uso: f(x, y) = x^2 - y^2  MODIFICAR AQUI LA FUNCION---
x_sym, y_sym = sympy.symbols('x y')
func_example_sym = x_sym**2-y_sym**2

P_extracted, q_extracted, r_extracted = get_quadratic_form_coeffs(func_example_sym, [x_sym, y_sym])

print(f"Función simbólica: {func_example_sym}")
print(f"\nMatriz P extraída:\n{P_extracted}")
print(f"Vector q extraído: {q_extracted}")
print(f"Escalar r extraído: {r_extracted}")

# --- Verificación usando compute_quadratic ---
print("\n--- Verificación con compute_quadratic ---")
x_test = np.array([1, 2])
original_value = float(func_example_sym.subs({x_sym: x_test[0], y_sym: x_test[1]}))
reconstructed_value = compute_quadratic(P_extracted, q_extracted, r_extracted, x_test)

print(f"Valor original de la función en x={x_test}: {original_value}")
print(f"Valor reconstruido con P, q, r en x={x_test}: {reconstructed_value}")
print(f"¿Coinciden los valores? {np.isclose(original_value, reconstructed_value)}")
```

### Obtener la Forma Cuadrática ($P$, $q$, $r$) de una Función Simbólica

Podemos usar `sympy` para descomponer una función cuadrática en sus componentes matriciales y vectoriales: $P$, $q$ y $r$, donde la función se expresa como $f(x) = x^T P x + q^T x + r$.

**solving the -dimension quadratic
system**

```python
import numpy as np

def solve_nd_quadratic(P, q, r=0):
    """
    Finds the minimizer of an N-D quadratic system,
    raises an error if there is no minimizer
    (P is not positive-definite).
    The quadratic function is assumed to be of the form f(x) = 0.5 * x^T P x + q^T x + r.
    Inputs:
    - P: A symmetric, positive-definite 2D numpy array (Hessian matrix).
    - q: A 1D numpy array (vector) for the linear term.
    - r: A scalar for the constant term (does not affect the minimizer).
    Output:
    - xstar: The minimizer, an (n x 1) vector.
    """
    # Check if P is positive-definite (all eigenvalues > 0)
    # np.linalg.eigvalsh returns eigenvalues for a symmetric matrix
    assert np.all(np.linalg.eigvalsh(P) > 0), "Matrix P must be positive-definite for a unique minimizer."

    # The minimizer x* for f(x) = 0.5 * x^T P x + q^T x + r is given by Px = -q, so x = -P^-1 * q
    return -np.linalg.solve(P, q)

#solution
P_matrix = np.array([[4, 1], [1, 2]]) # Let's rename P to P_matrix to avoid confusion with the function parameter
q_vector = np.array([3, 1]) # Let's rename q to q_vector

# solve the system
# If the quadratic form is f(x) = x^T A x + q^T x + r, then its Hessian is 2A.
# The solve_nd_quadratic function expects the Hessian (P) as its first argument.
# So, if P_matrix represents 'A', we pass 2.0 * P_matrix as the Hessian.
xstar_exact = solve_nd_quadratic(2.0 * P_matrix, q_vector)
print(f"Minimizer x*: [{xstar_exact[0]:.4f}, {xstar_exact[1]:.4f}]")
```

**Solucion por medio de la forma cuadratica**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

# Local implementation of plot_contour, replacing the external dependency
def plot_contour(func, x_range, y_range, ax, num_points=100):
    x_min, x_max = x_range
    y_min, y_max = y_range
    X = np.linspace(x_min, x_max, num_points)
    Y = np.linspace(y_min, y_max, num_points)
    X_grid, Y_grid = np.meshgrid(X, Y)

    # Evaluate the function Z at each point (x, y)
    Z = np.zeros_like(X_grid, dtype=float)
    for i in range(X_grid.shape[0]):
        for j in range(X_grid.shape[1]):
            # func expects a 1D array [x_coord, y_coord]
            Z[i, j] = func(np.array([X_grid[i, j], Y_grid[i, j]]))

    contour_plot = ax.contourf(X_grid, Y_grid, Z, levels=50, cmap='viridis')
    return contour_plot

# The function for the contour plot: f(x) = x^T P_matrix x + q_vector^T x
# The minimizer xstar_exact was found for this function (considering r=0).
f_for_contour = lambda x_coords : compute_quadratic(P_matrix, q_vector, 0, x_coords)

# make contour plot
fig, ax = plt.subplots(1, 1, figsize=(8, 6)) # Set a good figure size
plot_object = plot_contour(f_for_contour, [-5, 5] , [-5, 5], ax) # Adjust range for better visualization

# Plot the minimizer
ax.plot(xstar_exact[0], xstar_exact[1], 'r*', markersize=12, label='Minimizer')

# Add titles and labels
ax.set_title('Contour Plot of Quadratic Function and its Minimizer')
ax.set_xlabel('x_1')
ax.set_ylabel('x_2')
ax.legend()

# Add a colorbar
fig.colorbar(plot_object, ax=ax, format='%.0f')

plt.show()
```

### Calcular el Gradiente de una Función Simbólica

El gradiente de una función multivariable es un vector de sus derivadas parciales. Si tienes una función $f(x_1, x_2, \dots, x_n)$, su gradiente se denota como $\nabla f$ y es igual a:

$\nabla f = \left( \frac{\partial f}{\partial x_1}, \frac{\partial f}{\partial x_2}, \dots, \frac{\partial f}{\partial x_n} \right)$

```python
import sympy
import numpy as np

def compute_gradient(sym_func, sym_vars):
    """
    Calcula el gradiente simbólico de una función multivariable.

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        list: Una lista de expresiones simbólicas que representan las componentes del gradiente.
    """
    gradient = [sympy.diff(sym_func, var) for var in sym_vars]
    return gradient

# --- Ejemplo 1: f(x, y) = 3x^2 + y^2 ---
x_sym, y_sym = sympy.symbols('x y')
func1_sym = 3*x_sym**2 + y_sym**2

gradient1 = compute_gradient(func1_sym, [x_sym, y_sym])

print(f"Función simbólica 1: {func1_sym}")
print(f"Gradiente de la función 1: {gradient1}")

# --- Ejemplo 2: f(x, y, z) = x*y^2 - z*x^3 ---
x_sym, y_sym, z_sym = sympy.symbols('x y z')
# Se corrige para usar x_sym en lugar de la variable global 'x' de numpy
func2_sym = x_sym*y_sym**2 - z_sym*x_sym**3

gradient2 = compute_gradient(func2_sym, [x_sym, y_sym, z_sym])

print(f"\nFunción simbólica 2: {func2_sym}")
print(f"Gradiente de la función 2: {gradient2}")


# Opcional: Evaluar el gradiente en un punto específico
print("\n--- Evaluación numérica del Gradiente ---")

# Para el Ejemplo 1 en x=1, y=2
gradient1_eval = [expr.subs({x_sym: 1, y_sym: 2}) for expr in gradient1]
print(f"Gradiente de la función 1 en (1, 2): {gradient1_eval}")

# Para el Ejemplo 2 en x=1, y=2, z=3
gradient2_eval = [expr.subs({x_sym: 1, y_sym: 2, z_sym: 3}) for expr in gradient2]
print(f"Gradiente de la función 2 en (1, 2, 3): {gradient2_eval}")
```

### Calcular la Matriz Jacobiana de un Conjunto de Funciones Simbólicas

La **matriz Jacobiana** de un sistema de $m$ funciones de $n$ variables, $f: \mathbb{R}^n \to \mathbb{R}^m$, es una matriz que contiene todas las primeras derivadas parciales de cada función con respecto a cada variable. Se denota como $J_f$ o $Jf$.

Si $f = (f_1, f_2, \dots, f_m)$ y $x = (x_1, x_2, \dots, x_n)$, entonces la matriz Jacobiana $J_f$ es:

$$ J_f = \begin{pmatrix} \frac{\partial f_1}{\partial x_1} & \frac{\partial f_1}{\partial x_2} & \cdots & \frac{\partial f_1}{\partial x_n} \\ \frac{\partial f_2}{\partial x_1} & \frac{\partial f_2}{\partial x_2} & \cdots & \frac{\partial f_2}{\partial x_n} \\ \vdots & \vdots & \ddots & \vdots \\ \frac{\partial f_m}{\partial x_1} & \frac{\partial f_m}{\partial x_2} & \cdots & \frac{\partial f_m}{\partial x_n} \end{pmatrix} $$

En `sympy`, podemos calcular esto fácilmente utilizando la función `sympy.Matrix.jacobian()`.

```python
import sympy
import numpy as np

def compute_jacobian(sym_funcs, sym_vars):
    """
    Calcula la matriz Jacobiana simbólica para un conjunto de funciones.

    Args:
        sym_funcs (list): Una lista de expresiones simbólicas, donde cada una es una función.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        sympy.Matrix: La matriz Jacobiana simbólica.
    """
    # Convertir la lista de funciones a una matriz columna de SymPy
    f_vector = sympy.Matrix(sym_funcs)
    # Calcular el Jacobiano
    jacobian_matrix = f_vector.jacobian(sym_vars)
    return jacobian_matrix

# --- Ejemplo de uso: Sistema de dos funciones con dos variables ---
x_sym, y_sym = sympy.symbols('x y')

# Definimos las funciones f1 y f2
f1 = x_sym**2 * y_sym
f2 = sympy.sin(x_sym) + y_sym**3

# Creamos una lista con las funciones
functions_list = [f1, f2]
# Creamos una lista con las variables
variables_list = [x_sym, y_sym]

jacobian_example = compute_jacobian(functions_list, variables_list)

print(f"Funciones: {functions_list}")
print(f"Variables: {variables_list}")
print(f"\nMatriz Jacobiana:\n{jacobian_example}")

# --- Ejemplo de evaluación numérica del Jacobiano en un punto específico ---
print("\n--- Evaluación numérica del Jacobiano ---")

# Punto en el que queremos evaluar el Jacobiano
point_eval = {x_sym: 1, y_sym: 2}

# Para evaluar, podemos convertir la matriz SymPy a una función numérica
# o simplemente sustituir los valores directamente en la matriz SymPy.

# Opción 1: Sustituir directamente en la matriz SymPy
jacobian_numeric = jacobian_example.subs(point_eval)
print(f"Jacobiano evaluado en {point_eval}:\n{jacobian_numeric}")

# Para obtener un array de NumPy:
jacobian_numpy = np.array(jacobian_numeric, dtype=float)
print(f"Jacobiano evaluado (NumPy array):\n{jacobian_numpy}")

# Opción 2: Usar lambdify para crear una función Python/NumPy callable
from sympy.utilities.lambdify import lambdify

jacobian_callable = lambdify(variables_list, jacobian_example, 'numpy')
jacobian_numeric_from_callable = jacobian_callable(1, 2)
print(f"\nJacobiano evaluado con lambdify en {point_eval}:\n{jacobian_numeric_from_callable}")
```

### Verificación de la Condición de Convexidad de Primer Orden

Para una función `f` diferenciable, la condición de convexidad de primer orden establece que `f` es convexa si y solo si, para cualquier `x` e `y` en el dominio de `f`:

$$ f(y) \ge f(x) + \nabla f(x)^T (y - x) $$

Geométricamente, esto significa que la función siempre se encuentra por encima de su plano tangente en cualquier punto `x`.

```python
import sympy
import numpy as np

# Reutilizamos la función compute_gradient definida anteriormente
def compute_gradient(sym_func, sym_vars):
    gradient = [sympy.diff(sym_func, var) for var in sym_vars]
    return gradient

def check_first_order_convexity_condition(sym_func, sym_vars, x_point_vals, y_point_vals):
    """
    Verifica la condición de convexidad de primer orden: f(y) >= f(x) + grad(f(x))^T * (y-x).

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función f.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].
        x_point_vals (np.array): Un array numpy con los valores del punto x.
        y_point_vals (np.array): Un array numpy con los valores del punto y.

    Returns:
        bool: True si la condición se cumple, False en caso contrario.
        tuple: (f_y, rhs), los valores de f(y) y el lado derecho de la desigualdad.
    """
    if len(x_point_vals) != len(sym_vars) or len(y_point_vals) != len(sym_vars):
        raise ValueError("El número de valores en x_point_vals y y_point_vals debe coincidir con el número de variables simbólicas.")

    # 1. Calcular f(y)
    subs_y = {sym_vars[i]: y_point_vals[i] for i in range(len(sym_vars))}
    f_y = float(sym_func.subs(subs_y))

    # 2. Calcular f(x)
    subs_x = {sym_vars[i]: x_point_vals[i] for i in range(len(sym_vars))}
    f_x = float(sym_func.subs(subs_x))

    # 3. Calcular el gradiente de f(x)
    grad_f_sym = compute_gradient(sym_func, sym_vars)

    # 4. Evaluar el gradiente en el punto x
    grad_f_x = np.array([float(expr.subs(subs_x)) for expr in grad_f_sym])

    # 5. Calcular (y - x)
    diff_y_x = y_point_vals - x_point_vals

    # 6. Calcular grad(f(x))^T * (y-x)
    dot_product = np.dot(grad_f_x, diff_y_x)

    # 7. Calcular el lado derecho de la desigualdad
    rhs = f_x + dot_product

    print(f"\nFunción: {sym_func}")
    print(f"Punto x: {x_point_vals}, f(x)={f_x:.4f}")
    print(f"Punto y: {y_point_vals}, f(y)={f_y:.4f}")
    print(f"Gradiente de f en x: {grad_f_x}")
    print(f"Vector (y - x): {diff_y_x}")
    print(f"Lado Derecho (f(x) + grad(f(x))^T (y-x)): {rhs:.4f}")

    is_condition_met = f_y >= rhs
    print(f"¿Se cumple f(y) >= f(x) + grad(f(x))^T (y-x)? {is_condition_met}")

    # Pequeña tolerancia para comparaciones de punto flotante
    # return f_y >= rhs - 1e-9, (f_y, rhs)
    return is_condition_met, (f_y, rhs)

# --- Ejemplo 1: Función Convexa (f(x, y) = x^2 + y^2) ---
x_sym, y_sym = sympy.symbols('x y')
convex_func = x_sym**2 + y_sym**2

print("\n--- Verificación para Función Convexa (f(x, y) = x^2 + y^2) ---")
x_point1 = np.array([1.0, 2.0])
y_point1 = np.array([2.0, 3.0])
check_first_order_convexity_condition(convex_func, [x_sym, y_sym], x_point1, y_point1)

x_point2 = np.array([-2.0, 0.0])
y_point2 = np.array([0.0, -1.0])
check_first_order_convexity_condition(convex_func, [x_sym, y_sym], x_point2, y_point2)


# --- Ejemplo 2: Función No Convexa (f(x, y) = x^2 - y^2) ---
non_convex_func = x_sym**2 - y_sym**2

print("\n--- Verificación para Función No Convexa (f(x, y) = x^2 - y^2) ---")
x_point3 = np.array([0, 0])
y_point3 = np.array([1.0, 0.0])
check_first_order_convexity_condition(non_convex_func, [x_sym, y_sym], x_point3, y_point3)

x_point4 = np.array([0, 0.0])
y_point4 = np.array([0.0, -1.0])
check_first_order_convexity_condition(non_convex_func, [x_sym, y_sym], x_point4, y_point4)
```

### Cálculo de Autovalores (Eigenvalues) de una Matriz

Los **autovalores** son escalares especiales asociados con un sistema de ecuaciones lineales (o una matriz) que son de particular importancia en el análisis matricial y sus aplicaciones. Representan las 'direcciones' a lo largo de las cuales una transformación lineal actúa simplemente estirando o encogiendo un vector, sin cambiar su dirección.

Para una matriz cuadrada $A$, un número $\lambda$ se llama autovalor si existe un vector $v$ no nulo (llamado autovector) tal que:

$$ Av = \lambda v $$

En Python, `numpy.linalg.eigvals` se utiliza para calcular los autovalores de una matriz cuadrada.

```python
import numpy as np

# Define una matriz cuadrada de ejemplo
# Puedes modificar esta matriz a tu gusto
A = np.array([
    [1, 0],
    [0,-1]
])

print(f"Matriz cuadrada A:\n{A}")

# Calcula los autovalores
# numpy.linalg.eigvals devuelve un array de los autovalores de la matriz.
# Si la matriz es simétrica, se puede usar np.linalg.eigvalsh para mayor eficiencia y estabilidad numérica.
# En este caso, la matriz A es simétrica, por lo que eigvalsh sería una buena opción.

# Para matrices generales (no necesariamente simétricas):
# eigenvalues = np.linalg.eigvals(A)

# Para matrices simétricas (como la de ejemplo, es preferible usar eigvalsh):
eigenvalues = np.linalg.eigvalsh(A)


print(f"\nAutovalores de la matriz A: {eigenvalues}")
```

### Cálculo de la Matriz Hessiana de una Función Simbólica

La **matriz Hessiana** de una función escalar $f(x_1, x_2, \dots, x_n)$ de $n$ variables es una matriz cuadrada de segundas derivadas parciales. Se utiliza para describir la curvatura local de una función de valor escalar y es fundamental en problemas de optimización para determinar si un punto crítico es un mínimo local, un máximo local o un punto de silla.

Para una función $f(x_1, x_2, \dots, x_n)$, la matriz Hessiana $H(f)$ se define como:

$$ H(f)_{ij} = \frac{\partial^2 f}{\partial x_i \partial x_j} $$

En `sympy`, puedes calcular fácilmente la matriz Hessiana utilizando `sympy.hessian(función, variables)`.

```python
import sympy
import numpy as np

def compute_hessian(sym_func, sym_vars):
    """
    Calcula la matriz Hessiana simbólica de una función multivariable.

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        sympy.Matrix: La matriz Hessiana simbólica.
    """
    hessian_matrix = sympy.hessian(sym_func, sym_vars)
    return hessian_matrix

# --- Ejemplo 1: f(x, y) = x^2 + y^2 --- (Función Convexa)
x_sym, y_sym = sympy.symbols('x y')
func1_sym = x_sym**2 + y_sym**2

hessian1 = compute_hessian(func1_sym, [x_sym, y_sym])

print(f"Función simbólica 1: {func1_sym}")
print(f"Matriz Hessiana de la función 1:\n{hessian1}")

# --- Ejemplo 2: f(x, y) = 3*x^2 + 2*x*y + y^2 + 4*x + 5*y + 6 --- (Función Cuadrática con término cruzado)
x_sym, y_sym = sympy.symbols('x y')
func2_sym = 3*x_sym**2 + 2*x_sym*y_sym + y_sym**2 + 4*x_sym + 5*y_sym + 6

hessian2 = compute_hessian(func2_sym, [x_sym, y_sym])

print(f"\nFunción simbólica 2: {func2_sym}")
print(f"Matriz Hessiana de la función 2:\n{hessian2}")

# --- Ejemplo 3: f(x, y) = sin(x) + cos(y) --- (Función no lineal)
x_sym, y_sym = sympy.symbols('x y')
func3_sym = sympy.sin(x_sym) + sympy.cos(y_sym)

hessian3 = compute_hessian(func3_sym, [x_sym, y_sym])

print(f"\nFunción simbólica 3: {func3_sym}")
print(f"Matriz Hessiana de la función 3:\n{hessian3}")

# --- Evaluación numérica del Hessiano en un punto específico (Opcional) ---
print("\n--- Evaluación numérica del Hessiano ---")

# Para el Ejemplo 1 en x=1, y=2
point_eval1 = {x_sym: 1, y_sym: 2}
hessian1_numeric = hessian1.subs(point_eval1)
print(f"Hessiano de la función 1 en {point_eval1}:\n{hessian1_numeric}")

# Para el Ejemplo 3 en x=pi/2, y=pi (para sin(x) + cos(y))
point_eval3 = {x_sym: sympy.pi/2, y_sym: sympy.pi}
hessian3_numeric = hessian3.subs(point_eval3)
print(f"Hessiano de la función 3 en {point_eval3}:\n{hessian3_numeric}")

# Convertir a NumPy array (si es necesario)
hessian3_numpy = np.array(hessian3_numeric, dtype=float)
print(f"Hessiano de la función 3 en {point_eval3} (NumPy array):\n{hessian3_numpy}")
```

# Análisis Completo de Funciones Cuadráticas: Extracción, Convexidad y Optimización

A continuación, se presenta un conjunto de funciones y un flujo de trabajo para analizar una función simbólica, extraer su forma cuadrática, determinar su convexidad, encontrar su minimizador (si existe) y visualizarlo.

```python
import sympy
import numpy as np
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
def compute_quadratic(P, q, r, x):
    """
    Evaluates a quadratic function f(x) = x^T P x + q^T x + r.
    Inputs:
    - P: A 2D numpy array (matrix) for the quadratic term.
    - q: A 1D numpy array (vector) for the linear term.
    - r: A scalar for the constant term.
    - x: A 1D numpy array (vector) at which the quadratic function is evaluated.

    Returns:
    - f(x): The scalar value of the quadratic function.
    """
    # Calculate the quadratic term: x^T * P * x
    quadratic_term = np.dot(x.T, np.dot(P, x))
    # Calculate the linear term: q^T * x
    linear_term = np.dot(q.T, x)
    # Combine all terms
    return quadratic_term + linear_term + r
def get_quadratic_form_coeffs(sym_func, sym_vars):
    """
    Extrae la matriz P, el vector q y el escalar r de una función cuadrática simbólica.
    Asume que la función es de la forma f(x) = x^T P x + q^T x + r.

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función cuadrática.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        tuple: (P, q, r) como numpy arrays o escalar.
    """
    num_vars = len(sym_vars)

    # 1. Calcular r (término constante)
    # Evaluar la función en el origen (todas las variables = 0)
    subs_dict = {var: 0 for var in sym_vars}
    r = float(sym_func.subs(subs_dict))

    # 2. Calcular q (vector lineal)
    # El gradiente de f(x) en x=0 nos da q
    gradient = [sympy.diff(sym_func, var) for var in sym_vars]
    q_list = [float(expr.subs(subs_dict)) for expr in gradient]
    q = np.array(q_list)

    # 3. Calcular P (matriz cuadrática)
    # La matriz P es 0.5 * la Matriz Hessiana de f(x)
    hessian = sympy.hessian(sym_func, sym_vars)
    P_sympy = 0.5 * hessian
    P = np.array(P_sympy.tolist(), dtype=float) # Convertir la matriz de sympy a numpy

    return P, q, r

def compute_gradient(sym_func, sym_vars):
    """
    Calcula el gradiente simbólico de una función multivariable.

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        list: Una lista de expresiones simbólicas que representan las componentes del gradiente.
    """
    gradient = [sympy.diff(sym_func, var) for var in sym_vars]
    return gradient

def compute_hessian(sym_func, sym_vars):
    """
    Calcula la matriz Hessiana simbólica de una función multivariable.

    Args:
        sym_func (sympy.Expr): La expresión simbólica de la función.
        sym_vars (list): Una lista de símbolos de sympy que representan las variables [x1, x2, ...].

    Returns:
        sympy.Matrix: La matriz Hessiana simbólica.
    """
    hessian_matrix = sympy.hessian(sym_func, sym_vars)
    return hessian_matrix



# --- Ejemplo de uso: f(x, y) = x^2 + y^2  MODIFICAR AQUI LA FUNCION---
x_sym, y_sym = sympy.symbols('x y')
func_example_sym = x_sym**2+y_sym**2-2


P_extracted, q_extracted, r_extracted = get_quadratic_form_coeffs(func_example_sym, [x_sym, y_sym])


# Para matrices simétricas (como la de ejemplo, es preferible usar eigvalsh):
eigenvalues = np.linalg.eigvalsh(P_extracted)


gradient1 = compute_gradient(func_example_sym, [x_sym, y_sym])

P_inversa = np.linalg.inv(P_extracted)
x_minimo=-1*P_inversa*q_extracted


hessian1 = compute_hessian(func_example_sym, [x_sym, y_sym])



print(f"Función simbólica: {func_example_sym}")
print(f"\nMatriz P extraída:\n{P_extracted}")
print(f"\nAutovalores de la matriz P: {eigenvalues}")
print(f"Vector q extraído: {q_extracted}")
print(f"Escalar r extraído: {r_extracted}")

# --- Verificación usando compute_quadratic ---
print("\n--- Verificación con compute_quadratic ---")
x_test = np.array([1, 2])
original_value = float(func_example_sym.subs({x_sym: x_test[0], y_sym: x_test[1]}))
reconstructed_value = compute_quadratic(P_extracted, q_extracted, r_extracted, x_test)


print(f"Valor original de la función en x={x_test}: {original_value}")
print(f"Valor reconstruido con P, q, r en x={x_test}: {reconstructed_value}")
print(f"¿Coinciden los valores? {np.isclose(original_value, reconstructed_value)}")


print(f"gradiente: {gradient1}")
print(f"minomo: {x_minimo}")


print(f"Matriz Hessiana de la función 1:\n{hessian1}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. DEFINIR EL PROBLEMA (mismo que antes)
# ============================================================

# Matriz de datos y observaciones
A = np.array([[1, 2],
              [2, 4],
              [3, 6]])
y = np.array([5, 10, 15])
c = 0.1  # regularización

# Calcular matrices fijas para el gradiente
ATA = A.T @ A
ATy = A.T @ y
I = np.eye(A.shape[1])
P = 2 * (ATA + c * I)
q = -2 * ATy

# Definir la función de costo
def f(x):
    return x @ P @ x / 2 + q @ x + y @ y

# Definir el gradiente
def grad(x):
    return P @ x + q

# Definir la Hessiana (para comprobar convexidad)
def hessian():
    return P

# ============================================================
# 2. IMPLEMENTAR EL ALGORITMO DE DESCENSO GENERAL
# ============================================================

def descent_method(f, grad, x0, alpha_strategy, tol=1e-6, max_iter=1000):
    """
    Algoritmo general de descenso.

    Parámetros:
    - f: función objetivo
    - grad: gradiente de f
    - x0: punto inicial
    - alpha_strategy: diccionario con la estrategia para alpha
        ej: {'type': 'constant', 'value': 0.01}
            {'type': 'decaying', 'a': 1}
            {'type': 'exact'} (búsqueda lineal exacta)
    - tol: tolerancia para el criterio de parada
    - max_iter: número máximo de iteraciones
    """
    x = x0.copy()
    history = [x.copy()]
    f_history = [f(x)]

    for k in range(max_iter):
        # 1. Calcular el gradiente
        g = grad(x)

        # 2. Comprobar criterio de parada (gradiente pequeño)
        if np.linalg.norm(g) < tol:
            print(f"Convergencia alcanzada en iteración {k}. Norma del gradiente = {np.linalg.norm(g):.2e}")
            break

        # 3. Determinar la dirección de descenso (gradiente negativo)
        delta_x = -g

        # 4. Elegir el tamaño de paso según la estrategia
        if alpha_strategy['type'] == 'constant':
            alpha = alpha_strategy['value']

        elif alpha_strategy['type'] == 'decaying':
            a = alpha_strategy.get('a', 1)
            alpha = 1 / (a + k)

        elif alpha_strategy['type'] == 'exact':
            # Búsqueda lineal exacta: minimizar phi(alpha) = f(x + alpha * delta_x)
            # Como f es cuadrática, phi(alpha) es una parábola, podemos encontrar su mínimo exacto.
            # phi(alpha) = 0.5 * (x + alpha*d)^T P (x + alpha*d) + q^T (x + alpha*d) + cte
            # La derivada es: phi'(alpha) = (x + alpha*d)^T P d + q^T d
            # Igualando a cero: alpha = -(x^T P d + q^T d) / (d^T P d)
            d = delta_x
            num = x @ P @ d + q @ d
            den = d @ P @ d
            if den > 0:
                alpha = -num / den
                alpha = max(0, alpha)  # alpha debe ser >= 0
            else:
                alpha = 0.01  # fallback

        else:
            raise ValueError("Estrategia de alpha no reconocida")

        # 5. Actualizar el punto
        x_new = x + alpha * delta_x

        # 6. Guardar historial
        history.append(x_new.copy())
        f_history.append(f(x_new))

        # 7. Preparar siguiente iteración
        x = x_new

    return np.array(history), np.array(f_history)

# ============================================================
# 3. EJECUTAR CON DIFERENTES ESTRATEGIAS
# ============================================================

x0 = np.array([0.0, 0.0])  # punto inicial

# Definir las estrategias
strategies = [
    {'name': 'Constante (alpha=0.01)', 'params': {'type': 'constant', 'value': 0.01}},
    {'name': 'Constante (alpha=0.05)', 'params': {'type': 'constant', 'value': 0.05}},
    {'name': 'Decreciente (a=1)', 'params': {'type': 'decaying', 'a': 1}},
    {'name': 'Búsqueda lineal exacta', 'params': {'type': 'exact'}},
]

# Ejecutar cada estrategia y guardar resultados
results = {}
for strat in strategies:
    print(f"\n{'='*60}")
    print(f"Ejecutando: {strat['name']}")
    print('='*60)
    hist, f_hist = descent_method(f, grad, x0, strat['params'], tol=1e-8, max_iter=200)
    results[strat['name']] = (hist, f_hist)
    print(f"Solución final: {hist[-1]}")
    print(f"Valor final de la función: {f_hist[-1]:.6f}")

# ============================================================
# 4. COMPARAR CON LA SOLUCIÓN ANALÍTICA
# ============================================================

# Solución analítica (Ecuaciones Normales)
x_star = np.linalg.solve(ATA + c * I, ATy)
print(f"\n{'='*60}")
print("SOLUCIÓN ANALÍTICA (REFERENCIA)")
print('='*60)
print(f"x* = {x_star}")
print(f"f(x*) = {f(x_star):.6f}")

# ============================================================
# 5. VISUALIZACIÓN DE LA CONVERGENCIA
# ============================================================

# Graficar la convergencia del valor de la función
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
for name, (hist, f_hist) in results.items():
    plt.semilogy(f_hist, label=name)
plt.axhline(y=f(x_star), color='r', linestyle='--', label='Óptimo analítico')
plt.xlabel('Iteración')
plt.ylabel('f(x) (escala log)')
plt.title('Convergencia del valor de la función')
plt.legend()
plt.grid(True, alpha=0.3)

# Graficar la trayectoria en el espacio de parámetros
plt.subplot(1, 2, 2)
for name, (hist, f_hist) in results.items():
    plt.plot(hist[:, 0], hist[:, 1], marker='.', markersize=3, label=name)
plt.plot(x_star[0], x_star[1], 'rx', markersize=12, label='Óptimo analítico')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Trayectoria en el espacio de parámetros')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================
# 6. ANÁLISIS ADICIONAL: EFECTO DEL TAMAÑO DE PASO
# ============================================================

print("\n" + "="*60)
print("ANÁLISIS DEL ERROR FINAL")
print("="*60)

for name, (hist, f_hist) in results.items():
    error_x = np.linalg.norm(hist[-1] - x_star)
    error_f = abs(f_hist[-1] - f(x_star))
    print(f"{name}:")
    print(f"  Error en x: {error_x:.2e}")
    print(f"  Error en f: {error_f:.2e}")
    print(f"  Iteraciones: {len(f_hist)}")
```

### Demostración del Cálculo Analítico de la Tasa de Aprendizaje Óptima (Exact Line Search)

El método de gradiente descendente para una función cuadrática $f(x) = \frac{1}{2} x^T P x + q^T x + r$ tiene una tasa de aprendizaje óptima $\alpha^*$ que puede ser calculada analíticamente. Esta $\alpha^*$ minimiza la función en la dirección de descenso $d_k = -\nabla f(x_k)$ y se define como:

$\alpha_k^* = -\frac{d_k^T \nabla f(x_k)}{d_k^T P d_k}$

El siguiente código demuestra esta estrategia, utilizando las definiciones de función y el algoritmo de descenso ya presentes en el notebook, específicamente para el problema de regresión Ridge definido anteriormente.

```python
import numpy as np
import matplotlib.pyplot as plt

# Reutilizamos las definiciones de A, y, c del ejemplo de Ridge Regression
A = np.array([[1, 2],
              [2, 4],
              [3, 6]])
y = np.array([5, 10, 15])
c = 0.1  # regularización

# Calculamos los parámetros P y q para la función cuadrática f(x) = 0.5 * x^T P x + q^T x + r
ATA = A.T @ A
ATy = A.T @ y
I = np.eye(A.shape[1])
P_quad = 2 * (ATA + c * I) # Matriz P de la forma cuadrática 0.5 * x^T P x + q^T x + r
q_linear = -2 * ATy      # Vector q de la forma cuadrática
r_constant = y @ y       # Término constante r

# Definir la función de costo (f(x) = 0.5 * x^T P_quad x + q_linear^T x + r_constant)
def f(x):
    return 0.5 * x @ P_quad @ x + q_linear @ x + r_constant

# Definir el gradiente (grad(x) = P_quad @ x + q_linear)
def grad(x):
    return P_quad @ x + q_linear

# Reutilizamos la función de descenso general (adaptada para usar P_quad)
def descent_method_with_exact_line_search(f, grad, x0, P_matrix, alpha_strategy, tol=1e-6, max_iter=1000):
    x = x0.copy()
    history = [x.copy()]
    f_history = [f(x)]

    for k in range(max_iter):
        g = grad(x)

        if np.linalg.norm(g) < tol:
            print(f"Convergencia alcanzada en iteración {k}. Norma del gradiente = {np.linalg.norm(g):.2e}")
            break

        delta_x = -g

        if alpha_strategy['type'] == 'exact_quad':
            # Implementación de la fórmula exacta para funciones cuadráticas:
            # alpha_k^* = - (d_k^T * grad(f(x_k))) / (d_k^T * P * d_k)
            d = delta_x
            # El numerador es d_k^T * grad(f(x_k)) = d_k^T * (P_quad @ x + q_linear)
            num = d @ (P_matrix @ x + q_linear)
            # El denominador es d_k^T * P_quad * d_k
            den = d @ P_matrix @ d

            if den > 0:
                alpha = -num / den
                alpha = max(0, alpha)  # alpha debe ser >= 0
            else:
                alpha = 0.01  # Fallback si el denominador es no positivo (caso raro para P_quad positiva definida)
        else:
            raise ValueError("Estrategia de alpha no reconocida para esta demostración.")

        x_new = x + alpha * delta_x
        history.append(x_new.copy())
        f_history.append(f(x_new))
        x = x_new

    return np.array(history), np.array(f_history)

# Punto inicial
x0 = np.array([0.0, 0.0])

# Definir la estrategia de búsqueda lineal exacta
exact_strategy = {'type': 'exact_quad'}

print("=" * 60)
print("Ejecutando Gradiente Descendente con Exact Line Search (Cuadrático)")
print("=" * 60)
hist_exact, f_hist_exact = descent_method_with_exact_line_search(f, grad, x0, P_quad, exact_strategy, tol=1e-8, max_iter=50)

print(f"Solución final: {hist_exact[-1]}")
print(f"Valor final de la función: {f_hist_exact[-1]:.6f}")

# Solución analítica para comparación
x_star_analytical = np.linalg.solve(P_quad, -q_linear)
print(f"\n{'='*60}")
print("SOLUCIÓN ANALÍTICA (REFERENCIA)")
print('='*60)
print(f"x* = {x_star_analytical}")
print(f"f(x*) = {f(x_star_analytical):.6f}")

# ============================================================
# Visualización de la Convergencia
# ============================================================

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.semilogy(f_hist_exact, label='Búsqueda lineal exacta')
plt.axhline(y=f(x_star_analytical), color='r', linestyle='--', label='Óptimo analítico')
plt.xlabel('Iteración')
plt.ylabel('f(x) (escala log)')
plt.title('Convergencia del valor de la función')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(hist_exact[:, 0], hist_exact[:, 1], marker='.', markersize=8, label='Búsqueda lineal exacta')
plt.plot(x_star_analytical[0], x_star_analytical[1], 'rx', markersize=12, label='Óptimo analítico')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Trayectoria en el espacio de parámetros')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

def quadratic_exact_line_search(P, q, x, Dx):
    """
    Encuentra el tamaño de paso exacto que minimiza un sistema cuadrático
    en un punto dado x para una dirección de búsqueda Dx.

    La función cuadrática es: f(x) = 0.5 * x^T P x + q^T x + r (la constante r no afecta al paso).

    Fórmula del paso exacto:
    alpha* = - (Dx^T P x + Dx^T q) / (Dx^T P Dx)

    Inputs:
    - P: matriz del sistema cuadrático (n x n, simétrica)
    - q: vector de coeficientes lineales (n x 1)
    - x: punto actual (n x 1)
    - Dx: dirección de búsqueda (n x 1)

    Output:
    - t: tamaño de paso óptimo (alpha*)
    """

    # 1. Calcular (Dx^T P)
    PDx = Dx.T @ P  # vector fila de tamaño (1 x n)

    # 2. Calcular el denominador: (Dx^T P Dx) -> debe ser > 0 para que sea una dirección de descenso
    DxTPDx = PDx @ Dx  # escalar

    # Si el denominador es <= 0, la dirección no es de descenso o el sistema no es convexo
    if DxTPDx <= 0:
        # En este caso, devolvemos un valor pequeño para evitar división por cero
        # o lanzamos una advertencia
        print(f"Advertencia: Dx^T P Dx = {DxTPDx:.6f} <= 0. Usando alpha = 0.01")
        return 0.01

    # 3. Calcular (Dx^T q)
    DxTq = Dx.T @ q  # escalar

    # 4. Calcular (Dx^T P x)
    DxTPx = PDx @ x  # escalar

    # 5. Calcular el paso exacto
    t = -(DxTPx + DxTq) / DxTPDx

    # 6. Asegurar que alpha sea no negativo (por si hay errores numéricos)
    if t < 0:
        print(f"Advertencia: alpha = {t:.6f} es negativo. Forzando a 0.")
        t = 0.0

    return t


# ============================================================
# EJEMPLO DE USO
# ============================================================

if __name__ == "__main__":
    # Definir un sistema cuadrático de prueba
    # f(x) = 0.5 * x^T P x + q^T x + r
    P = np.array([[3.0, 1.0],
                  [1.0, 2.0]])
    q = np.array([-5.0, -3.0])  # equivalente a -b en la forma 0.5 x^T A x - b^T x

    # Punto actual y dirección de búsqueda (por ejemplo, el residual)
    x = np.array([10.0, -5.0])
    Dx = np.array([-1.0, 2.0])  # dirección de descenso (negativo del gradiente, por ejemplo)

    # Calcular el paso exacto
    alpha = quadratic_exact_line_search(P, q, x, Dx)

    print("="*60)
    print("EJEMPLO DE BÚSQUEDA LINEAL EXACTA")
    print("="*60)
    print(f"P = \n{P}")
    print(f"q = {q}")
    print(f"x = {x}")
    print(f"Dx = {Dx}")
    print(f"alpha* = {alpha:.6f}")

    # Verificar: evaluar f(x + alpha * Dx)
    def f(x):
        return 0.5 * x @ P @ x + q @ x

    f_current = f(x)
    f_new = f(x + alpha * Dx)

    print(f"f(x) = {f_current:.6f}")
    print(f"f(x + alpha*Dx) = {f_new:.6f}")
    print(f"Reducción: {f_current - f_new:.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. FUNCIONES AUXILIARES
# ============================================================

def quadratic_exact_line_search(P, q, Dx, x):
    """
    Encuentra el tamaño de paso exacto para un sistema cuadrático.
    f(x) = 0.5 * x^T P x + q^T x + r
    """
    PDx = Dx.T @ P
    DxTPDx = PDx @ Dx
    DxTq = Dx.T @ q
    DxTPx = PDx @ x
    t = -(DxTPx + DxTq) / DxTPDx
    return t

def solve_nd_quadratic(P, q):
    """
    Resuelve el sistema cuadrático analíticamente: x* = -P^{-1} q
    """
    return np.linalg.solve(P, -q)

def plot_contour(f, x_range, y_range, ax, levels=30):
    """
    Dibuja las curvas de nivel de la función f.
    """
    x1 = np.linspace(x_range[0], x_range[1], 50)
    x2 = np.linspace(y_range[0], y_range[1], 50)
    X1, X2 = np.meshgrid(x1, x2)
    Z = np.zeros_like(X1)
    for i in range(X1.shape[0]):
        for j in range(X1.shape[1]):
            Z[i, j] = f([X1[i, j], X2[i, j]])
    contour = ax.contourf(X1, X2, Z, levels=levels, cmap='viridis', alpha=0.7)
    return contour

# ============================================================
# 2. ALGORITMO DE DESCENSO DEL GRADIENTE
# ============================================================

def gradient_descent_quadratic(P, q, x0, epsilon=1e-4, trace=False):
    """
    Descenso del gradiente para sistemas cuadráticos.

    Inputs:
    - P, q: términos del sistema cuadrático f(x) = 0.5 * x^T P x + q^T x + r
    - x0: punto de partida
    - epsilon: tolerancia para el criterio de parada
    - trace: (bool) si es True, devuelve la trayectoria y el número de pasos

    Outputs:
    - xstar: el mínimo encontrado
    - trajectory: trayectoria de las iteraciones (si trace=True)
    - n_steps: número de pasos antes de terminar (si trace=True)
    """
    x = x0.copy()
    trajectory = [x.copy()]  # Lista para guardar la trayectoria
    n_steps = 0

    while n_steps < 1000:
        # 1. Calcular la dirección de descenso (negativo del gradiente)
        Dx = -P @ x - q

        # 2. Criterio de parada: si el gradiente es muy pequeño
        if np.linalg.norm(Dx) < epsilon:
            break

        # 3. Búsqueda lineal exacta: calcular el tamaño de paso óptimo
        t = quadratic_exact_line_search(P, q, Dx, x)

        # 4. Actualizar el punto
        x = x + t * Dx

        # 5. Guardar la trayectoria
        trajectory.append(x.copy())
        n_steps += 1

    # Convertir la trayectoria a un array numpy
    trajectory = np.array(trajectory)

    if trace:
        return x, trajectory, n_steps
    else:
        return x

# ============================================================
# 3. DEFINIR EL PROBLEMA Y EJECUTAR
# ============================================================

# Parámetros del problema
gamma = 10.0

# Forma cuadrática: f(x) = 0.5 * (x1^2 + gamma * x2^2)
# En forma matricial: f(x) = 0.5 * x^T P x + q^T x + r
P = np.array([[1.0, 0.0],
              [0.0, gamma]])
q = np.array([0.0, 0.0])  # No hay término lineal

# Punto inicial
x0 = np.array([2.0, 1.0])

# Tolerancia
epsilon = 1e-8

# ============================================================
# 4. EJECUTAR EL DESCENSO DEL GRADIENTE
# ============================================================

print("="*60)
print("DESCENSO DEL GRADIENTE PARA SISTEMA CUADRÁTICO")
print("="*60)
print(f"Función: f(x) = 0.5 * (x1^2 + {gamma} * x2^2)")
print(f"Punto inicial: x0 = {x0}")
print(f"Tolerancia: epsilon = {epsilon}")
print()

# Ejecutar el algoritmo con trace=True para obtener la trayectoria
xstar_gd, trajectory, n_steps = gradient_descent_quadratic(P, q, x0, epsilon=epsilon, trace=True)

# Solución analítica (exacta)
xstar_exact = solve_nd_quadratic(P, q)

print(f"Solución GD encontrada: x* = {xstar_gd}")
print(f"Solución analítica: x* = {xstar_exact}")
print(f"Número de iteraciones: {n_steps}")
print(f"Error respecto a la solución exacta: {np.linalg.norm(xstar_gd - xstar_exact):.2e}")

# ============================================================
# 5. VISUALIZACIÓN
# ============================================================

# Definir la función para las curvas de nivel
def f(x):
    return 0.5 * np.array(x) @ P @ np.array(x) + q @ np.array(x)

# Crear la figura
fig, ax = plt.subplots(1, 1, figsize=(10, 8))

# Dibujar curvas de nivel
plot_obj = plot_contour(f, [-4, 4], [-4, 4], ax)
fig.colorbar(plot_obj, ax=ax, label='f(x)')

# Marcar la solución encontrada por GD
ax.plot(xstar_gd[0], xstar_gd[1], 'r*', markersize=12, label='Solución GD')

# Marcar la solución analítica
ax.plot(xstar_exact[0], xstar_exact[1], 'bo', markersize=10, label='Solución analítica')

# Marcar el punto inicial
ax.plot(x0[0], x0[1], 'go', markersize=10, label='Punto inicial')

# Dibujar la trayectoria
ax.plot(trajectory[:, 0], trajectory[:, 1], 'r.-', linewidth=1.5, markersize=4, label='Trayectoria GD')

# Etiquetas y leyenda
ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.set_title(f'Descenso del gradiente con búsqueda lineal exacta\nγ = {gamma}, Iteraciones = {n_steps}')
ax.legend()
ax.grid(True, alpha=0.3)
ax.axis('equal')

plt.tight_layout()
plt.show()

# ============================================================
# 6. ANÁLISIS ADICIONAL (para responder la pregunta)
# ============================================================

print("\n" + "="*60)
print("ANÁLISIS DE LA TRAYECTORIA")
print("="*60)

# Mostrar los primeros pasos de la trayectoria
print("\nPrimeros 5 pasos de la trayectoria:")
for i in range(min(5, len(trajectory))):
    print(f"  Iteración {i}: x = {trajectory[i]}")

# Calcular la norma del gradiente en cada paso
grad_norms = []
for x in trajectory:
    grad = P @ x + q
    grad_norms.append(np.linalg.norm(grad))

print(f"\nNorma del gradiente inicial: {grad_norms[0]:.2e}")
print(f"Norma del gradiente final: {grad_norms[-1]:.2e}")

# Calcular los valores de la función en la trayectoria
f_values = [f(x) for x in trajectory]
print(f"Valor inicial de f: {f_values[0]:.6f}")
print(f"Valor final de f: {f_values[-1]:.6f}")

print("\n" + "="*60)
print("FIN DEL EJEMPLO")
print("="*60)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. FUNCIONES AUXILIARES (necesarias para el código)
# ============================================================

def quadratic_exact_line_search(P, q, Dx, x):
    """
    Encuentra el tamaño de paso exacto para un sistema cuadrático.
    f(x) = 0.5 * x^T P x + q^T x + r
    """
    PDx = Dx.T @ P
    DxTPDx = PDx @ Dx
    DxTq = Dx.T @ q
    DxTPx = PDx @ x
    t = -(DxTPx + DxTq) / DxTPDx
    return t

def solve_nd_quadratic(P, q):
    """
    Resuelve el sistema cuadrático analíticamente: x* = -P^{-1} q
    """
    return np.linalg.solve(P, -q)

def gradient_descent_quadratic(P, q, x0, epsilon=1e-4, trace=False):
    """
    Descenso del gradiente para sistemas cuadráticos.

    Inputs:
    - P, q: términos del sistema cuadrático f(x) = 0.5 * x^T P x + q^T x + r
    - x0: punto de partida
    - epsilon: tolerancia para el criterio de parada
    - trace: (bool) si es True, devuelve la trayectoria y el número de pasos

    Outputs:
    - xstar: el mínimo encontrado
    - trajectory: trayectoria de las iteraciones (si trace=True)
    - n_steps: número de pasos antes de terminar (si trace=True)
    """
    x = x0.copy()
    trajectory = [x.copy()]
    n_steps = 0

    while n_steps < 1000:
        Dx = -P @ x - q
        if np.linalg.norm(Dx) < epsilon:
            break
        t = quadratic_exact_line_search(P, q, Dx, x)
        x = x + t * Dx
        trajectory.append(x.copy())
        n_steps += 1

    trajectory = np.array(trajectory)

    if trace:
        return x, trajectory, n_steps
    else:
        return x

# ============================================================
# 2. DEFINIR EL PROBLEMA
# ============================================================

gamma = 10.0

# Forma cuadrática: f(x) = 0.5 * (x1^2 + gamma * x2^2)
P = np.array([[1.0, 0.0],
              [0.0, gamma]])
q = np.array([0.0, 0.0])

# Punto inicial
x0 = np.array([2.0, 1.0])

# Tolerancia
epsilon = 1e-8

# ============================================================
# 3. EJECUTAR EL DESCENSO DEL GRADIENTE
# ============================================================

print("="*60)
print("DIAGRAMA DE CONVERGENCIA - DESCENSO DEL GRADIENTE")
print("="*60)

# Ejecutar el algoritmo con trace=True para obtener la trayectoria
xstar_gd, trajectory, n_steps_gd = gradient_descent_quadratic(P, q, x0, epsilon=epsilon, trace=True)

# Solución analítica (exacta)
xstar_exact = solve_nd_quadratic(P, q)

print(f"Solución GD encontrada: x* = {xstar_gd}")
print(f"Solución analítica: x* = {xstar_exact}")
print(f"Número de iteraciones: {n_steps_gd}")

# ============================================================
# 4. COMPLETAR: FUNCIÓN OBJETIVO Y CÁLCULO DEL ERROR
# ============================================================

# Definir la función objetivo
def f(x):
    """f(x) = 0.5 * (x1^2 + gamma * x2^2)"""
    return 0.5 * (x[0]**2 + gamma * x[1]**2)

# Calcular el valor mínimo exacto (en este caso es 0)
p_star = f(xstar_exact)

# Calcular el error en cada iteración: |f(x^(k)) - f(x*)|
error = [abs(f(x) - p_star) for x in trajectory]

# ============================================================
# 5. COMPLETAR: GRÁFICO DE CONVERGENCIA
# ============================================================

fig, ax = plt.subplots(1, 1, figsize=(10, 6))

# Graficar el error en escala logarítmica
ax.semilogy(error, 'b.-', linewidth=1.5, markersize=6, label='Error |f(x) - f(x*)|')

# Añadir una línea de referencia para mostrar la tendencia lineal
# (en escala log, la convergencia lineal aparece como una línea recta)

# Configurar el gráfico
ax.set_xlabel('Número de iteraciones (k)', fontsize=12)
ax.set_ylabel('Error |f(x) - f(x*)| (escala log)', fontsize=12)
ax.set_title(f'Diagrama de convergencia del Descenso del Gradiente\nγ = {gamma}, Iteraciones = {n_steps_gd}', fontsize=14)
ax.grid(True, alpha=0.3, which='both')
ax.legend()

plt.tight_layout()
plt.show()

# ============================================================
# 6. ANÁLISIS ADICIONAL PARA CONFIRMAR LA TEORÍA
# ============================================================

print("\n" + "="*60)
print("ANÁLISIS DE LA TASA DE CONVERGENCIA")
print("="*60)

# Calcular el número de condición de la matriz P
eigenvals = np.linalg.eigvals(P)
lambda_min = np.min(eigenvals)
lambda_max = np.max(eigenvals)
kappa = lambda_max / lambda_min

print(f"Autovalores de P: {eigenvals}")
print(f"Autovalor mínimo (λ1): {lambda_min:.6f}")
print(f"Autovalor máximo (λn): {lambda_max:.6f}")
print(f"Número de condición (κ): {kappa:.6f}")

# Factor de convergencia teórico
factor = (kappa - 1) / (kappa + 1)
print(f"Factor de convergencia teórico: {factor:.6f}")

print("\nTasa de convergencia teórica:")
print(f"  Iteración 1: error ≤ {factor**2:.4f} * error inicial")
print(f"  Iteración 2: error ≤ {factor**4:.4f} * error inicial")
print(f"  Iteración 3: error ≤ {factor**6:.4f} * error inicial")

# Verificar la relación teórica en la práctica
print("\nVerificación práctica:")
for k in range(min(5, len(error))):
    if k == 0:
        print(f"  Iteración {k}: error = {error[k]:.6f}")
    else:
        # Calcular la reducción real del error
        reduction = error[k] / error[k-1]
        print(f"  Iteración {k}: error = {error[k]:.6f}, reducción = {reduction:.4f} (teórico: {factor**2:.4f})")

# ============================================================
# 7. COMPARACIÓN CON DIFERENTES VALORES DE GAMMA
# ============================================================

print("\n" + "="*60)
print("COMPARACIÓN CON DIFERENTES VALORES DE γ")
print("="*60)

gammas = [1.0, 10.0, 100.0, 1000.0]
colors = ['b', 'g', 'r', 'c']

fig2, ax2 = plt.subplots(1, 1, figsize=(10, 6))

for i, g in enumerate(gammas):
    P_g = np.array([[1.0, 0.0], [0.0, g]])
    xstar_g, trajectory_g, n_steps_g = gradient_descent_quadratic(P_g, q, x0, epsilon=epsilon, trace=True)

    # Definir la función para este gamma
    def f_g(x, gamma=g):
        return 0.5 * (x[0]**2 + gamma * x[1]**2)

    p_star_g = f_g(xstar_g)
    error_g = [abs(f_g(x) - p_star_g) for x in trajectory_g]

    # Calcular el número de condición
    eigenvals_g = np.linalg.eigvals(P_g)
    kappa_g = np.max(eigenvals_g) / np.min(eigenvals_g)

    # Graficar
    ax2.semilogy(error_g, color=colors[i], linewidth=1.5,
                 label=f'γ = {g}, κ = {kappa_g:.2f}, iter = {n_steps_g}')

    print(f"γ = {g}: κ = {kappa_g:.2f}, iteraciones = {n_steps_g}")

ax2.set_xlabel('Número de iteraciones (k)', fontsize=12)
ax2.set_ylabel('Error |f(x) - f(x*)| (escala log)', fontsize=12)
ax2.set_title('Comparación de convergencia para diferentes valores de γ', fontsize=14)
ax2.grid(True, alpha=0.3, which='both')
ax2.legend()

plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("FIN DEL EJEMPLO")
print("="*60)
```
